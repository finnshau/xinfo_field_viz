# X-Info Catalog Pivot Table

Group plays by the five selected formation, personnel, front, coverage, and box-count variables. The result includes the number of plays in each combination.

```python
from pathlib import Path
import pandas as pd

catalog_path = Path("data/raw/xinfo_catalog2025_9_24.csv")
catalog = pd.read_csv(catalog_path, low_memory=False)

pivot_variables = [
    "offensive_formation",
    "offensive_personnel",
    "defensive_front",
    "coverage_concept_desc",
    "men_in_box",
]

pivot_table = (
    catalog.pivot_table(
        index=pivot_variables,
        values="play_unique_id",
        aggfunc="count",
    )
    .rename(columns={"play_unique_id": "play_count"})
    .reset_index()
    .sort_values("play_count", ascending=False)
    .reset_index(drop=True)
)

display(pivot_table)
```
